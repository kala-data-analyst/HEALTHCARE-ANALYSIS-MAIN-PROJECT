STEP 5 — CHECK OUTLIERS

```python
# STEP 5 - CHECK OUTLIERS

numeric_columns = df_clean.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    print(col, ":", outliers, "outliers")
```

NO NEED TO DELETE —not yet. An unusually high admission deposit can be a genuine value, not a data-entry error. We should inspect it before treating it.

```python
plt.figure(figsize=(10, 5))

# Plot histogram with dynamic teal palette and custom KDE line
ax = sns.histplot(
    df_clean["Age"],
    bins=20,
    kde=True,
    color="#2b5c8f",
    edgecolor="white",
    alpha=0.8,
)

# Color the KDE line distinctly
if ax.lines:
    ax.lines[0].set_color("#e63946")
    ax.lines[0].set_linewidth(2.5)

# Chart styling
plt.title(
    "Distribution of Patient Age",
    fontsize=14,
    fontweight="bold",
    pad=15,
    color="#1d3557",
)
plt.xlabel("Age (Years)", fontsize=11, fontweight="bold")
plt.ylabel("Number of Patients", fontsize=11, fontweight="bold")

# Add average reference line with label
mean_age = df_clean["Age"].mean()
plt.axvline(
    mean_age, color="#e63946", linestyle="--", linewidth=1.8, label=f"Mean Age: {mean_age:.1f}"
)

# Remove top and right borders for clarity
sns.despine()
plt.legend(frameon=True, facecolor="white")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean aesthetic style
sns.set_theme(style="whitegrid", font_scale=1.05)

plt.figure(figsize=(10, 6))

# Plot categorical bar plot using a colorful gradient palette
ax = sns.countplot(
    data=df_clean,
    x="Department",
    order=df_clean["Department"].value_counts().index,
    palette="viridis",
    hue="Department",
    legend=False,
)

# Add exact numeric count labels on top of each bar for presentation clarity
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f"{int(height):,}",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold",
            color="#2b2b2b",
            xytext=(0, 3),
            textcoords="offset points",
        )

# Styling and Titles
plt.title(
    "Patient Distribution by Department",
    fontsize=14,
    fontweight="bold",
    pad=15,
    color="#1d3557",
)
plt.xlabel("Department", fontsize=11, fontweight="bold")
plt.ylabel("Number of Patients", fontsize=11, fontweight="bold")
plt.xticks(rotation=30, ha="right", fontsize=10)

# Remove top and right borders
sns.despine()

plt.tight_layout()
plt.show()
```
